# Research-area persistence benchmark - revision-robustness Colab runbook

**Drive layout** (create `MyDrive/who-inherits/` and place):

```
MyDrive/who-inherits/
  colab_bundle.zip      <- uploaded by you (contains code, frozen data,
                           partial cache, partial results, notebooks)
  .openalex_key         <- one line, plain text (never enters the bundle,
                           never printed by any notebook)
  results/robustness/   <- created by the notebooks (checkpoint syncs)
  DONE_fetch.flag  DONE_theta.flag  DONE_rgcn.flag  DONE_topk.flag
```

**Run order and runtimes**

| notebook | runtime | when |
|---|---|---|
| `01_fetch_r5` | **free CPU runtime** | start immediately (network-bound; ~18k of 89k authors remain) |
| `02_theta_sweep_r1` | A100 runtime (uses its CPUs) | start in parallel with 01 |
| `03_rgcn_r3` | **A100** | after 02 (or first - it owns the GPU) |
| `04_topk_r6` | A100 runtime's CPUs or a second CPU runtime | only after `DONE_fetch.flag` exists |

**Done =** four `DONE_*.flag` files on Drive. Then sync
`MyDrive/who-inherits/results/robustness/` back into the local repo's
`results/robustness/` and run `python code/r7_merge_colab.py`.

Every notebook is safe to re-run after a disconnect: it pulls the synced
results from Drive first and skips completed units (a lost session costs at
most one unit: one theta cell, one grid config, or <=1,000 fetched authors).
